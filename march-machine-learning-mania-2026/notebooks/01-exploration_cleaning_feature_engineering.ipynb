```python
import polars as pl
import polars.selectors as cs
from src import exploration_cleaning_methods as ecm
```

**Men's NCCA data**

```python
m_regular_season_detailed_lf: pl.LazyFrame = pl.scan_csv("../data/raw/MRegularSeasonDetailedResults.csv")
m_regular_season_detailed_lf.head(10).collect()
```

```python
m_tourney_detailed_lf: pl.LazyFrame = pl.scan_csv("../data/raw/MNCAATourneyDetailedResults.csv")
m_tourney_detailed_lf.head(10).collect()
```

```python
m_tourney_detailed_lf = m_tourney_detailed_lf.with_columns(cs.numeric().exclude("WLoc").cast(pl.Int16))
m_tourney_detailed_lf.head(10).collect()
```

```python
m_tourney_detailed_lf = ecm.create_team_season_profile(m_tourney_detailed_lf)
m_tourney_detailed_lf.head(10).collect()
```

```python
m_tourney_detailed_df: pl.DataFrame = m_tourney_detailed_lf.collect()
m_tourney_detailed_df.write_parquet("../data/proccessed/m_tourney_detailed.parquet")
```

```python
del m_tourney_detailed_df, m_tourney_detailed_lf
```

```python
m_tourney_seed_lf: pl.LazyFrame = pl.scan_csv("../data/raw/MNCAATourneySeeds.csv")
m_tourney_seed_lf.head(10).collect()
```

```python
m_tourney_seed_lf = m_tourney_seed_lf.with_columns(cs.numeric().exclude("Seed").cast(pl.Int16))
m_tourney_seed_lf.head(10).collect()
```

```python
m_tourney_seed_lf = ecm.clean_seed(m_tourney_seed_lf)
m_tourney_seed_lf.head(10).collect()
```

```python
m_golden_tabel_lf: pl.LazyFrame = ecm.merge_seed_with_regular(m_tourney_seed_lf, m_regular_season_detailed_lf)
m_golden_tabel_lf.head(10).collect()
```

```python
del m_regular_season_detailed_lf, m_tourney_seed_lf
```

```python
m_tourney_matchups_lf: pl.LazyFrame = pl.scan_csv("../data/raw/MNCAATourneyDetailedResults.csv")
m_tourney_matchups_lf = ecm.prepare_tourney_matchups(m_tourney_matchups_lf)
m_tourney_matchups_lf.head(10).collect()
```

```python
m_train_data_lf: pl.LazyFrame = ecm.finalize_training_data(m_tourney_matchups_lf, m_golden_tabel_lf)
m_train_data_lf.head(10).collect()
```

```python
del m_tourney_matchups_lf, m_golden_tabel_lf
```

```python
m_train_data_df: pl.DataFrame = m_train_data_lf.collect()
m_train_data_df.write_parquet("../data/proccessed/m_train_data.parquet")
```

```python
del m_train_data_df, m_train_data_lf
```
